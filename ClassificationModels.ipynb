```python
from sklearn.preprocessing import LabelEncoder
columns = ['sex','embarked','class','who','adult_male','embark_town','alive','alone']
le = LabelEncoder()
for col in columns:
  df[col] = le.fit_transform(df[col])
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[['age','fare']] = scaler.fit_transform(df[['age','fare']])
#fit_transform() returns a NumPy array rather than a DataFrame.
```

```python
df
```

```python
input = df.drop(columns=['survived'])
output = df['survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
Xtrain,Xtest,Ytrain,Ytest=train_test_split(input,output,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score
logestic_model = LogisticRegression()
```

```python
logestic_model.fit(Xtrain,Ytrain)
```

```python
prediction = logestic_model.predict(Xtest)
```

```python
r2_score(Ytest,prediction)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy_score(Ytest,prediction)
```

```python
confusion_matrix(Ytest,prediction)
```

```python
classification_report(Ytest,prediction)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors = 9)
knn.fit(Xtrain,Ytrain)
predict = knn.predict(Xtest)
```

```python
accuracy_score(Ytest,predict)
```

```python
confusion_matrix(Ytest,predict)
```

```python
classification_report(Ytest,predict)
```

```python
from sklearn.tree import DecisionTreeClassifier
tree = DecisionTreeClassifier()
tree.fit(Xtrain,Ytrain)
predict_tree = tree.predict(Xtest)


```

```python
tree
```

```python
r2_score(Ytest,predict_tree)
```

```python
acc = accuracy_score(Ytest,predict_tree)
info = classification_report(Ytest,predict_tree)
print(acc)
print(info)
```

```python
confusion_matrix(Ytest,predict_tree)
```
