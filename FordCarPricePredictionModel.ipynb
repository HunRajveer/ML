```python
x = df.drop(columns = 'price', axis =1)
y = df['price']
```

```python
x
```

```python
x_one_encoded = pd.get_dummies(x , columns = ['model','transmission','fuelType'])
```

```python
df['transmission'].value_counts()
```

```python
x_one_encoded.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
columns = ['model','transmission','fuelType']
xlabel = x
for i in columns:
    xlabel[i] = encoder.fit_transform(xlabel[i])
```

```python
xlabel
```

```python
from sklearn.preprocessing import StandardScaler
# we will only do for numeric cols which are present in xlabel
scaler = StandardScaler()
numeric_cols = ['year','mileage','tax','mpg','engineSize']
x_one_encoded[numeric_cols]  = scaler.fit_transform(x_one_encoded[numeric_cols])
```

```python
x_one_encoded
```

```python
numeric_cols1= ['model','year','transmission','mileage','fuelType','tax','mpg','engineSize']
xlabel[numeric_cols1] = scaler.fit_transform(xlabel[numeric_cols1])
xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import  r2_score
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x_one_encoded,y,test_size = 0.30)
```

```python
model = LinearRegression()
model.fit(xtrain,ytrain)
```

```python
prediction = model.predict(xtest)
prediction
```

```python
ytest
```

```python
rsquare = r2_score(ytest,prediction)
rsquare
```

```python
n = xtest.shape[0]
p = xtest.shape[1]
ads = 1-((1-rsquare)*(n-1))/(n-p-1)
ads
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(xlabel,y,test_size = 0.34)
```

```python
model = LinearRegression()
model.fit(xtrain,ytrain)
```

```python
predict = model.predict(xtest)
predict
```

```python
rsquare = r2_score(ytest,predict)
rsquare
```
